## 1. Python essencial

### Variáveis e tipos

Uma **variável** é um nome que guarda um valor. Os tipos que aparecem no curso são poucos: **número inteiro**, **número decimal**, **texto** e **booleano** (verdadeiro/falso).

```python
n_estados = 27          # inteiro (int)
taxa_media = 22.74      # decimal (float)
disciplina = "Estatística"   # texto (str)
tem_prova = True        # booleano (bool)

print(n_estados, taxa_media, disciplina, tem_prova)
print(type(taxa_media))   # type() diz o tipo de um valor
```

### Aritmética

Os operadores usuais. Repare em dois menos óbvios: `**` é potência e `/` sempre dá decimal.

```python
print(2 + 3)      # 5
print(10 / 4)     # 2.5  (divisão sempre devolve decimal)
print(10 // 4)    # 2    (divisão inteira, descarta o resto)
print(2 ** 3)     # 8    (2 elevado a 3)
print(16 ** 0.5)  # 4.0  (raiz quadrada = elevar a 1/2)
```

### `print()` e f-strings — mostrar resultados

Quase toda saída do livro usa **f-string**: um texto com `f` na frente, onde o que está entre `{ }` é substituído pelo valor. É assim que se mistura texto e número numa frase.

O `:.1f` depois do valor controla as casas decimais (aqui, 1 casa).

```python
media = 22.74

print(f"A taxa média é {media}")        # o valor cru
print(f"A taxa média é {media:.1f}")    # com 1 casa decimal
print(f"São {n_estados} estados no total.")
```

### Listas

Uma **lista** guarda vários valores em ordem. Você acessa um item pela **posição**, que começa em **0** (o primeiro item é `notas[0]`). `notas[-1]` é o último.

```python
notas = [8, 6, 9, 7, 10]

print(notas[0])     # primeiro item: 8
print(notas[-1])    # último item: 10
print(len(notas))   # quantos itens: 5
print(sum(notas))   # soma: 40
```

### Comparações e `if`

Uma comparação (`>`, `<`, `==`, `!=`) devolve `True` ou `False`. O `if` executa um bloco só quando a condição é verdadeira; o `else` cuida do resto. **A indentação (o recuo) marca o que está dentro do `if`.**

```python
media = 22.74

if media > 20:
    print("Taxa alta")
else:
    print("Taxa baixa")
```

### `for` — repetir

O `for` percorre uma lista, um item de cada vez. De novo, o recuo marca o corpo do laço.

```python
for nota in notas:
    print(f"nota: {nota}")
```

### Ler uma *list comprehension*

Você vai **ver** (não precisa escrever) uma forma compacta de criar uma lista a partir de outra. Leia assim: *"para cada `nota` em `notas`, calcule `nota * 2`"*.

```python
dobros = [nota * 2 for nota in notas]
print(dobros)   # [16, 12, 18, 14, 20]
```

## 2. As três bibliotecas da disciplina

Uma **biblioteca** é um conjunto de ferramentas prontas. `import ... as ...` carrega a biblioteca com um apelido curto. Estas três aparecem em quase todo capítulo.

```python
import numpy as np                 # contas e sorteios
import pandas as pd                # tabelas de dados
import matplotlib.pyplot as plt    # gráficos
```

### numpy — contas e a **semente**

O numpy faz contas sobre muitos números de uma vez. E sorteia números ao acaso — mas o livro **sempre fixa a semente** com `default_rng(42)`, para que o sorteio dê **sempre o mesmo resultado**. Isso é essencial: garante que o número que você vê na tela é o mesmo que está impresso no material.

```python
print(np.sqrt(16))                 # 4.0
print(np.mean([8, 6, 9, 7, 10]))   # 8.0

rng = np.random.default_rng(42)    # semente fixa = 42
print(rng.integers(0, 2, 10))      # 10 sorteios de 0 ou 1

# Rode esta célula DUAS vezes seguidas: o resultado da linha abaixo é sempre o mesmo,
# porque a semente reinicia o sorteio no mesmo ponto.
print(np.random.default_rng(42).integers(0, 2, 10))
```

### pandas — a tabela de dados

O pandas lê tabelas. Ele consegue ler um CSV **direto de um endereço na web** — e é assim que vamos carregar os dados do curso, que ficam no repositório do livro no GitHub.

Uma tabela inteira é um **DataFrame**; uma coluna só é uma **Series**.

```python
estado = pd.read_csv("https://raw.githubusercontent.com/BragaD/UnDF-Bases3-Estatistica-202602/main/dados/estados.csv")
estado.head()   # mostra as primeiras linhas
```

Para pegar **uma coluna**, use colchetes com o nome dela. Sobre uma coluna numérica, os resumos são métodos diretos: `.mean()`, `.median()`, `.std()`. E `.describe()` dá vários de uma vez.

```python
populacao = estado["Populacao"]   # uma Series (a coluna de população)

print(f"Média:   {populacao.mean():.0f}")
print(f"Mediana: {populacao.median():.0f}")
print(f"Desvio:  {populacao.std():.0f}")

estado["Taxa.Homicidios"].describe()
```

### matplotlib — um gráfico

O padrão do livro para um gráfico: cria-se a figura com `plt.subplots()`, desenha-se nela (aqui, um histograma com `ax.hist`), rotula-se o eixo, e `plt.show()` exibe.

```python
fig, ax = plt.subplots()
ax.hist(estado["Populacao"] / 1e6, bins=10, edgecolor="white")
ax.set_xlabel("População (milhões)")
ax.set_ylabel("Número de estados")
plt.show()
```
